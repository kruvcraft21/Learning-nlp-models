```python
# ...existing code...

# Bokeh: скаттеры с доверительными интервалами (Whisker)
from bokeh.io import output_notebook, show
from bokeh.plotting import figure
from bokeh.models import ColumnDataSource, HoverTool, Whisker, Div
from bokeh.layouts import gridplot, column
from bokeh.transform import linear_cmap
from bokeh.palettes import Category10

output_notebook()

# Берём те же поля, что и в seaborn-версии с интервалами
cols = [
    "metrics.total_chunks", "metrics.avg_chunk_length",
    "metrics.MRR", "metrics.MRR_lower", "metrics.MRR_upper",
    "metrics.MAP", "metrics.MAP_lower", "metrics.MAP_upper",
    "metrics.RECALL_AT_10", "metrics.RECALL_AT_10_lower", "metrics.RECALL_AT_10_upper",
    "metrics.HIT_AT_10", "metrics.HIT_AT_10_lower", "metrics.HIT_AT_10_upper",
]
df = runs[runs["params.model"] == "deepvk/USER2-base"][cols].copy()

# Приводим типы к числовым и очищаем NaN
for c in df.columns:
    df[c] = pd.to_numeric(df[c], errors="coerce")
df = df.dropna()

if df.empty:
    show(Div(text="<b>Нет данных для визуализации (после фильтра и очистки)</b>"))
else:
    src = ColumnDataSource(df)

    palette = Category10[10]
    color_map = linear_cmap(
        field_name="metrics.MRR", palette=palette,
        low=df["metrics.MRR"].min(), high=df["metrics.MRR"].max()
    )

    def make_scatter_with_ci(x_field, y_field, y_lower, y_upper, x_label, y_label, title):
        p = figure(height=360, width=500, tools="pan,wheel_zoom,box_zoom,reset,save", title=title)
        r = p.scatter(x=x_field, y=y_field, source=src, size=12,
                      line_color="black", line_width=0.7, fill_color=color_map, alpha=0.85)
        p.xaxis.axis_label = x_label
        p.yaxis.axis_label = y_label

        # Вертикальные error bars (доверительные интервалы по метрике)
        whisker = Whisker(source=src, base=x_field, upper=y_upper, lower=y_lower,
                          line_width=2, line_alpha=0.7)
        whisker.upper_head.size = 10
        whisker.lower_head.size = 10
        p.add_layout(whisker)

        p.add_tools(HoverTool(renderers=[r], tooltips=[
            ("x", f"@{x_field}"),
            ("y", f"@{y_field}{0.000}"),
            ("y_lower", f"@{y_lower}{0.000}"),
            ("y_upper", f"@{y_upper}{0.000}"),
            ("MRR", "@{metrics.MRR}{0.000}"),
            ("MAP", "@{metrics.MAP}{0.000}"),
            ("Recall@10", "@{metrics.RECALL_AT_10}{0.000}"),
            ("Hit@10", "@{metrics.HIT_AT_10}{0.000}")
        ]))
        return p

    # Метрики vs Количество чанков
    p_mrr_chunks = make_scatter_with_ci("metrics.total_chunks", "metrics.MRR",
                                        "metrics.MRR_lower", "metrics.MRR_upper",
                                        "Количество чанков", "MRR", "MRR vs Количество чанков")

    p_map_chunks = make_scatter_with_ci("metrics.total_chunks", "metrics.MAP",
                                        "metrics.MAP_lower", "metrics.MAP_upper",
                                        "Количество чанков", "MAP", "MAP vs Количество чанков")

    p_rec_chunks = make_scatter_with_ci("metrics.total_chunks", "metrics.RECALL_AT_10",
                                        "metrics.RECALL_AT_10_lower", "metrics.RECALL_AT_10_upper",
                                        "Количество чанков", "RECALL@10", "RECALL@10 vs Количество чанков")

    p_hit_chunks = make_scatter_with_ci("metrics.total_chunks", "metrics.HIT_AT_10",
                                        "metrics.HIT_AT_10_lower", "metrics.HIT_AT_10_upper",
                                        "Количество чанков", "HIT@10", "HIT@10 vs Количество чанков")

    # Метрики vs Средняя длина чанков
    p_mrr_len = make_scatter_with_ci("metrics.avg_chunk_length", "metrics.MRR",
                                     "metrics.MRR_lower", "metrics.MRR_upper",
                                     "Ср. длина чанков", "MRR", "MRR vs Ср. длина чанков")

    p_map_len = make_scatter_with_ci("metrics.avg_chunk_length", "metrics.MAP",
                                     "metrics.MAP_lower", "metrics.MAP_upper",
                                     "Ср. длина чанков", "MAP", "MAP vs Ср. длина чанков")

    p_rec_len = make_scatter_with_ci("metrics.avg_chunk_length", "metrics.RECALL_AT_10",
                                     "metrics.RECALL_AT_10_lower", "metrics.RECALL_AT_10_upper",
                                     "Ср. длина чанков", "RECALL@10", "RECALL@10 vs Ср. длина чанков")

    p_hit_len = make_scatter_with_ci("metrics.avg_chunk_length", "metrics.HIT_AT_10",
                                     "metrics.HIT_AT_10_lower", "metrics.HIT_AT_10_upper",
                                     "Ср. длина чанков", "HIT@10", "HIT@10 vs Ср. длина чанков")

    grid = gridplot([
        [p_mrr_chunks, p_map_chunks],
        [p_rec_chunks, p_hit_chunks],
        [p_mrr_len,    p_map_len],
        [p_rec_len,    p_hit_len],
    ], toolbar_location="right")

    show(column(Div(text="<b>Bokeh Scatter — метрики с доверительными интервалами</b>"), grid))
# ...existing code...
```
